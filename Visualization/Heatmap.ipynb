```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.DataFrame({
    'Math': [90, 85, 80],
    'Science': [70, 75, 80],
    'English': [85, 95, 90]
}, index=['Student1', 'Student2', 'Student3'])
```

```python
class Myheatmap:
  def __init__(self,data):
    self.data = data

  def myheatmap(self):
    sns.heatmap(data,
            annot=True,
            cmap='YlGnBu',
            linewidths=0.5,
            linecolor='black',
            fmt='.1f',
            square=True,
            cbar=True,
            vmin=0, vmax=100,
            xticklabels=True,
            yticklabels=True)
    plt.title("Student Scores Heatmap")
    plt.show()
myyheatmap = Myheatmap(data)
myyheatmap.myheatmap()
```

```python
Data = pd.DataFrame({
    'Math': [90, 85, 88, 92, 95],
    'Science': [89, 84, 86, 91, 94],
    'English': [78, 75, 80, 82, 85],
    'History': [88, 82, 85, 87, 90]
})
```

```python
class Mymyheatmap:
  def __init__(self,Data):
    self.Data = Data

  def mymyheatmap(self):
    matrix_Data = Data.corr()
    sns.heatmap(matrix_Data,
                annot=True,
                linewidths=0.5,
                linecolor="black",
                fmt=".2f",
                cmap='YlGnBu',
                square = True,
                vmin = 0 ,vmax = 1,
                xticklabels=True,
                yticklabels=True,
                )

    plt.title("Heatmap of Subject Score Correlations")
    plt.show()

mymheatmap = Mymyheatmap(Data)
mymheatmap.mymyheatmap()
```
