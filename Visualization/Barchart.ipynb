```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = ["Green","Blue","Red","Brown"]
y_1 = [20,30,40,10]
y_2 = [30,40,20,30]
colors = ["green","blue","red","brown"]
```

```python
class MyBarchart:
  def __init__(self,X,y_1,y_2,colors):
    self.X = X
    self.y_1 = y_1
    self.y_2 = y_2
    self.colors = colors

  def barchart(self):
    plt.bar(
        self.X,
        self.y_1,
        color = self.colors,
        width = 0.5,
        edgecolor = 'black',
        linewidth = 1,
        align='center',
        alpha=1,
        label='colors'
        )
    plt.title("Eye color in class",color ="red")
    plt.xlabel("Eyes Color")
    plt.ylabel("Number of student")
    plt.legend()

  def subplot(self):
    fig,ax = plt.subplots()
    x = np.arange(len(colors))
    width = 0.35
    bars1 = ax.bar(x-width/2,y_1,color="deepskyblue",width = width,label = "8-YEAR",edgecolor="black",alpha=0.9 )
    bars2 = ax.bar(x+width/2,y_2,color="mediumseagreen",width = width,label = "9-YEAR",edgecolor="black",alpha=0.9 )
    ax.set_xticks(x)
    ax.set_xticklabels(self.X)
    ax.set_ylabel("Student Counts",fontdict ={"fontsize" : 10})
    ax.set_xlabel("Types of Eye colors",fontdict ={"fontsize" : 10})
    ax.set_title("8/9 Eye colors by Years",fontdict ={"fontsize" : 15},pad=15)
    ax.legend()
    ax.bar_label(bars1, padding = 3)
    ax.bar_label(bars2, padding = 3)


mybarchart = MyBarchart(X,y_1,y_2,colors)
# mybarchart.barchart()
mybarchart.subplot()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
years = ['2011', '2012', '2013']
fruit_values = [750, 760, 780]
veg_values = [550, 540, 570]

x = np.arange(len(years))
width = 0.35

# Create figure and axes
fig, ax = plt.subplots()

# Plot grouped bars
bars1 = ax.bar(x - width/2, fruit_values, width, label='Fruit', color='deepskyblue')
bars2 = ax.bar(x + width/2, veg_values, width, label='Vegetables', color='mediumseagreen')

# Add labels, title, and grid
ax.set_xlabel('Year')
ax.set_ylabel('Amount')
ax.set_title('Fruit and Vegetable Sales by Year',pad = 10)
ax.set_xticks(x)
ax.set_xticklabels(years)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.6)

# Add value labels without loop
ax.bar_label(bars1, padding=3)
ax.bar_label(bars2, padding=3)

plt.tight_layout()
plt.show()
```
