```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
class Myboxplot:
  def __init__(self,X):
    self.X = X

  def myboxplot(self):
    plt.figure(figsize=(8, 6))
    plt.boxplot([self.X['Feature A'],self.X['Feature B'],self.X['Feature C']],labels=["Feature A","Feature B","Feature C"],showfliers=True,flierprops=dict(marker='x',color = 'red' , markersize = 8))
    plt.ylabel("Values")
    plt.grid(True)
    plt.show()

X = pd.DataFrame({'Feature A': [10, 12, 14, 15, 13, 10, 11, 16, 17, 19, 35],'Feature B': [22, 21, 23, 20, 19, 24, 25, 21, 22, 23, 5],'Feature C': [5, 7, 6, 8, 6, 5, 7, 9, 10, 6, 15]})

boxplot = Myboxplot(X)
boxplot.myboxplot()
```
