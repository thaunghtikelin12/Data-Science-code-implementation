```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
countries = ['USA', 'India', 'Japan']
gdp = [21, 3, 5]
life_expectancy = [79, 70, 84]
population = [330, 1400, 126]
```

```python
class Mybubblechart:
  def __init__(self,countries,gdp,life_expectancy,population):
    self.countries = countries
    self.gdp = gdp
    self.life_expectancy = life_expectancy
    self.population = [p*10 for p in population]

  def mybubble(self):
    plt.figure(figsize=(9,6))
    plt.scatter(self.gdp,
                self.life_expectancy,
                s = self.population,
                c=['blue', 'orange', 'green'],
                alpha=0.8,marker = 'o',
                edgecolors='black',
                linewidths=1.5)
    plt.title("Bubble chart : GDB & LIFE EXPECTANCY (BUBBLE_SIZE = POPULATION)")
    plt.xlabel("GDP")
    plt.ylabel(" LIFE EXPECTANCY ")
    plt.grid(True,linestyle='--')
    plt.show()

bubblechart = Mybubblechart(countries,gdp,life_expectancy,population)
bubblechart.mybubble()
```
